### What is a Pandas Dataframe?
Pandas is a data manipulation tool which is built on the Numpy package. Pandas' key data structure is the `dataframe`. A dataframe allows for the storage and manipulation of tabular data. It is a two-dimensional labelled data structure. 

<img src="https://raw.githubusercontent.com/Explore-AI/Pictures/master/Pandas_dataframe.jpg" width="900">

Basically, you could say that the Pandas dataframe consists of three main components: 

- the data, 
- the index and 
- the columns. 

Let's walk through some examples to gain an understanding of these components.

#### Creating a Dataframe
We can create a dataframe by calling the `DataFrame()` constructor. The main arguments in the constructor are the data, index and columns. The data that is passed in can be in the form of other data structures (lists, dictionaries or numpy arrays) or by loading in a file.

Let's start our examples by first importing the Pandas and numpy libraries:

```python
# Installing Pandas and Numpy using pip
!pip install pandas
!pip install numpy

# Importing Pandas
import pandas as pd
import numpy as np
```

#### Dataframes from other Data Structures

We now look at some examples on how to create a dataframe from various data structures in Python.

### Method 1: Creating Pandas DataFrame from list of lists

Using **lists**:

1. we need to create a list of lists (i.e—nested list) with the relevant data. 
2. We then also need to pass an index and column names.

```python
# Create list of lists containing data of Name and Age.
list_data = [['John', 25], ['James', 33], ['Joseph', 60]]

# Create index - Nationality (John from South Africa, James from Kenya, Joseph from Nigeria)
index = ['South Africa','Kenya','Nigeria']

# Create column names
columns = ['Name','Age']

# Create a Dataframe 
df_lol = pd.DataFrame(data=list_data, index=index, columns=columns)
df_lol
```


### Method 2: Creating Pandas DataFrame from Dictionaries

Using **dictionaries**, we need to create a dictionary with the relevant data. 

The keys should be the column names, while the values should be the data entries for that column. We then also need to pass an index. Note that because the keys account for the column names, we don't have to pass in an argument for columns.

```python
# Create dictionary containing data.
dictionary = {
    'Name':['John', 'James', 'Joseph'], 
    'Height in meters':[1.6, 1.5, 1.7], 
    'Age':[25, 33, 60]
}

# Create index - nationality.
index = ['South Africa','Kenya','Nigeria']

# Create dataframe by passing in data and index.
df_dict = pd.DataFrame(index=index, data=dictionary)
df_dict
```

### Method 3: Creating Pandas DataFrame using Numpy arrays.

Using **numpy arrays**, we need to create a numpy array with the relevant data. We then also need to pass an index (Nationality) and column names.

```python
# Create numpy array containing data.
array_data = np.array([['John', 1.6, 20], ['James', 1.5, 33], ['Joseph', 1.7, 42]])

# Create index - nationality.
index = ['South Africa','Kenya','Nigeria']

# Create column names.
columns = ['Name','Heigh','Age']

# Create dataframe by passing in data, index and columns.
df_array = pd.DataFrame(data=array_data, index=index, columns=columns)
df_array
```

### Method 4: Dataframes from other files / sources

The data argument can also be a **loaded file**. We will load the full csv file. Pandas makes it very easy to do this --> you simply load the data using the `read_csv()` function and pass in the full path of the file as a string.

Pandas will use the first row as columns so we don't need to pass column names. 

```python
# For this exercise, we'll use some football data (Premier League) from the FootyStats website.
# Credit: https://footystats.org/download-stats-csv

# Calling file path
url = 'https://github.com/Gabe-Maja/Python-Notes/blob/main/england-premier-league-players-2018-to-2019-stats.csv?raw=true'

# Read in csv and convert to DataFrame
df_footy = pd.read_csv(url)
```

```python
# Use the .head() function to look at the first 5 rows.
df_footy.head()
```

```python
# Use the .tail() function to look at the last 5 rows.
df_footy.tail()
```

```python
# Use the .describe() function to get an overview of the DataFrame.
df_footy.describe()
```

```python
# Use the .info() function to  print a concise summary of the DataFrame.
df_footy.info()
```

 | `.tail()` | Function to only look at the last 5 records of our data. |

### Accessing Dataframes

Accessing data within dataframes is not as straightforward as with the previous data structures. This can be done by index, by column, or by both. Let's work through these methods.

#### By Index
To access by index only in a dataframe we can use the `iloc` or `loc` functions with the indices in square brackets. The `iloc` function refers to the index location, so we pass in the number of the index, while the `loc` function refers to the name of the index, so we pass in the index name. Use slicing if you want more than one index. Eg:

* `dataframe.iloc[index i]` - returns series at index i
* `dataframe.iloc[index start: index end]` - returns dataframe from start to end (end not included)
* `dataframe.loc['index name']` - returns series of given index name

Let's look at a few examples:

```python
# Creating dataframe
data = [['John', 42, 1.7, 50], ['Joseph', 15, 1.5, 42], ['James',25, 1.5, 36], ['Jeff', 18, 1.4, 60], 
        ['Yuri', 30, 1.4, 55], ['Chad', 33, 1.7, 56]]

columns = ['Name', 'Age', 'Height', 'Weight in Kg']

index = ['South Africa', 'Zambia', 'Kenya', 'Swaziland', 'Ghana', 'Nigeria']

df_data = pd.DataFrame(data=data, columns=columns, index=index)
df_data
```

```python
# Select the 4th row using iloc[].
df_data.iloc[3]
```

```python
# Select rows 2 to 5 using iloc.
df_data.iloc[1:5]
```

```python
# Select South Africa using loc[].
df_data.loc['South Africa']
```

#### By Column
To access by column only we can simply call `dataframe['Column Name']`. If we want more than one column we input a list of column names inside the square brackets:

* `dataframe['Column Name']` - returns series of given column
* `dataframe[['Column 1', 'Column 2']]` - returns dataframe with the given columns

Let's look at examples.

```python
# Select the column 'Age'.
df_data['Age']
```

```python
df_data.Age
```

```python
# Select the columns 'Age' and 'Height'.
df_data[['Age', 'Height']]
```

#### By index and column
We can also select a subset of the dataframe using indices and columns in combination. 

```python
# Select the first 4 rows and first 2 columns - Rows first.
df_data.iloc[0:4, 0:2]
```

### When to use Dataframes
Unlike the Numpy array which is suited for storing and performing computations on homogeneous data (data of the same type), Pandas dataframes can accommodate heterogeneous data. This makes them the choice data structure for manipulating often messy data (e.g tabular data from spreadsheets, raw CSV files or SQL tables). 

We should use a Pandas dataframe if all of the following statements hold:

* We have 2-dimensional data (rows and columns)
* The data type is the same within a column
* We are interested in the index (rows) and column names
